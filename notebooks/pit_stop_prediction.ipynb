# F1 Pit Stop Prediction

This notebook demonstrates the F1 Pit Stop Prediction pipeline using a modularized project structure.

## 1. Import Dependencies

```python
import os
import sys
import joblib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))

from data.data import load_raw_data
from pit_stop_prediction.data_preprocessing import engineer_features
from pit_stop_prediction.model_evaluation import evaluate_predictions, plot_confusion_matrix, plot_roc_curve
```

## 2. Load Raw Data from CSV

```python
df = load_raw_data()
print(f"Raw dataset shape: {df.shape}")
df.head()
```

## 3. Feature Engineering

```python
df_engineered = engineer_features(df)
print(f"Engineered dataset shape: {df_engineered.shape}")
df_engineered[['Compound', 'Compound_Hardness', 'TyreLife', 'Tyre_Stress', 'Degradation_Rate']].head()
```

## 4. Load Pretrained Pipeline and Model

```python
preprocessor_path = '../models/scaler.pkl'
model_path = '../models/predict_pit_stop.pkl'

if os.path.exists(preprocessor_path) and os.path.exists(model_path):
    preprocessor = joblib.load(preprocessor_path)
    model = joblib.load(model_path)
    print("Model and Preprocessor pipeline loaded successfully!")
else:
    print("Pretrained artifacts not found. Please run the training script first.")
```

## 5. Evaluate and Visualize Results

```python
target = 'PitNextLap'
X = df_engineered.drop(columns=['id', target])
y = df_engineered[target]

X_transformed = preprocessor.transform(X)

y_pred = model.predict(X_transformed)
y_prob = model.predict_proba(X_transformed)[:, 1]

metrics = evaluate_predictions(y, y_pred, y_prob)

plot_confusion_matrix(y, y_pred)
plt.show()

plot_roc_curve(y, y_prob)
plt.show()
```
